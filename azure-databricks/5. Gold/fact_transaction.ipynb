```python
%sql
CREATE SCHEMA IF NOT EXISTS gold
```

```python
%sql
CREATE TABLE IF NOT EXISTS gold.fact_transactions(
    transaction_id STRING,
    scr_transaction_id STRING,
    encounter_id STRING,
    FK_patient_id STRING,
    FK_provider_id STRING,
    FK_dept_id STRING,
    icd_code STRING,
    procedure_code STRING,
    visit_type STRING,
    service_date DATE,
    paid_date DATE,
    amount DOUBLE,
    paid_amount DOUBLE,
    amount_type STRING,
    claim_id STRING,
    datasource STRING,
    refreshed_at TIMESTAMP

)
```

```python
%sql
TRUNCATE TABLE gold.fact_transactions
```

```python
%sql
INSERT INTO gold.fact_transactions
SELECT 
transactionid,
scr_transactionid,
encounterid,
concat(patientid,'-',datasource) AS FK_patient_id,
CASE 
WHEN datasource = 'hospital-a' THEN CONCAT('H1-',providerid) 
ELSE concat('H2-',providerid) 
END AS FK_provider_id,
concat(deptid,'-',datasource) AS FK_dept_id,
icdcode,
procedurecode,
visittype,
servicedate,
paiddate,
amount,
paidamount,
amounttype,
claimid,
datasource,
current_timestamp()
FROM silver.transactions
WHERE is_current = true AND is_quarantined = false
```
