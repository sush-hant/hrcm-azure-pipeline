```python
from pyspark.sql import SparkSession, functions as f

claims_df=spark.read.csv("/mnt/landing/claims/*.csv",header=True)

claims_df = claims_df.withColumn(
    "datasource",
    f.when(f.input_file_name().contains("hospital_a"), "hospital-a").when(f.input_file_name().contains("hospital_b"), "hospital-b")
     .otherwise(None)
)

display(claims_df)
```

```python
claims_df.write.format("parquet").mode("overwrite").save("/mnt/bronze/claims/")
```
