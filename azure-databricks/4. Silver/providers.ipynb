```python
%sql
CREATE SCHEMA IF NOT EXISTS silver
```

```python
%sql
CREATE TABLE IF NOT EXISTS silver.providers(
  ProviderID string,
  firstname string,
  lastname string,
  specialization string,
  deptid string,
  npi long,
  datasource string,
  is_qurantined boolean
)
USING DELTA
```

```python
%sql
INSERT INTO silver.providers
SELECT 
  DISTINCT ProviderID,
  FirstName,
  LastName,
  Specialization,
  DeptID,
  CAST(NPI AS INT) NPI,
  datasource,
  CASE  
    WHEN providerid IS NULL OR DeptId IS NULL THEN TRUE
    ELSE FALSE
  END is_qurantined
FROM providers
```
